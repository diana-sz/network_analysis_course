```python
from numpy import random
import matplotlib.pyplot as plt

# For a large network_size (N) and a low edge probability (p), the binomial distribution
# is near identical to the poisson distribution, such that the 
# average degree (<k>) is equal to N*p 

# Set a default fontsize for all figures 
plt.rcParams.update({'font.size': 16})

def plot_binomial_poisson(network_size,edge_prob,no_samples):
    """Overlayed histograms of a binomial and the corresponding poisson distribution"""
    ax = plt.subplot(1, 1, 1)
    average_deg = network_size * edge_prob
    binomial_samples = random.binomial(n=network_size, p=edge_prob, size=no_samples)
    poisson_samples = random.poisson(lam=average_deg, size=no_samples)
    hist_binomial = plt.hist(binomial_samples, bins=100, alpha=0.5, density=True, label="binomial")
    hist_poisson = plt.hist(poisson_samples, bins=100, alpha=0.5, density=True, label="poisson")
    plt.text(0.05, 0.9, 'N: {}\np: {}'.format(network_size,edge_prob), transform=ax.transAxes)
    plt.xlabel("Edges (k)")
    plt.ylabel("Probability that a node has k edges ($p_k$)")
    plt.legend(loc='upper right')
    plt.tight_layout()

# Example 1
plt.figure(figsize=(6,6))
plot_binomial_poisson(100,0.5,10000)
plt.show()
plt.close()

# Example 2
plt.figure(figsize=(6,6))
plot_binomial_poisson(10000,0.001,10000)
plt.show()
plt.close()
```
