````python
import networkx as nx
import numpy as np
import random as rd
import warnings
````

````python
# Generate a network 
my_network = nx.scale_free_graph(100)
#my_network = nx.erdos_renyi_graph(500,0.3)
    
# We want to append to these lists to store information
all_random_degrees = [] # degrees of the random nodes 
all_neigh_avg_degrees = [] # average degrees of the neighbors
    
for i in range(1000):
    random_node = rd.choice(list(my_network.nodes()))
    random_degree = my_network.degree(random_node)
    all_random_degrees.append(random_degree)
    neigh_deg = [my_network.degree(n) for n in my_network.neighbors(random_node)]
    avg_neigh_deg = np.nanmean(neigh_deg)
    all_neigh_avg_degrees.append(avg_neigh_deg)

average_node = np.nanmean(all_random_degrees)
average_neighbors = np.nanmean(all_neigh_avg_degrees)
# Take nanmean b/c some nodes may have no neighbors
print("The average degree of the average node is {}".format(average_node))
print("The average degree of the neighbors is {}".format(average_neighbors))
````

NOTE: the `np.nanmean()` function might throw a warning \"mean of empty slice\". It's ok to leave this warning as is. Otherwise, you can turn it off by embedding the following code:
```
with warnings.catch_warnings():
    warnings.simplefilter(\"ignore\", category=RuntimeWarning)
    avg_neigh_deg = np.nanmean(neigh_deg)
```
