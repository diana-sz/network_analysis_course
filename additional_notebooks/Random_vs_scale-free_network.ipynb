```python
import networkx as nx
import matplotlib.pyplot as plt
```

### Generate a scale-free and a random network

```python
# Generating a network according to the Barabasi-Albert model (scale-free network)
ba_network = nx.barabasi_albert_graph(1000, 10, seed=33)

# Generate a random network that contains the same number of nodes and 
# the same number of edges as the scale-free network
random_network = nx.gnm_random_graph(ba_network.number_of_nodes(), 
                                     ba_network.number_of_edges(), 
                                     seed=33)
```

### Degree distribution of the scale-free network

```python
# Get the degree frequency where the degree is the index of the list
degree_freq = nx.degree_histogram(ba_network)
degrees = range(len(degree_freq))

# Plot the degrees against the degree frequencies 
plt.bar(degrees,degree_freq)
plt.xlabel("Degrees")
plt.ylabel("Count")
plt.show()
plt.close()
```

```python
# plot on log scale
plt.plot(degrees, degree_freq, 'o' )
plt.yscale('log')
plt.xscale('log')
plt.xlabel("Degrees")
plt.ylabel("Count")
plt.show()
plt.close()
```

### Degree distribution of the random network

```python
# Get the degree frequency where the degree is the index of the list
degree_freq2 = nx.degree_histogram(random_network)
degrees2 = range(len(degree_freq2))

# Plot the degrees against the degree frequencies 
plt.bar(degrees2, degree_freq2)
plt.xlabel("Degrees")
plt.ylabel("Count")
plt.show()
plt.close()
```

```python
# plot on log scale
plt.plot(degrees2, degree_freq2, 'o' )
plt.yscale('log')
plt.xscale('log')
plt.xlabel("Degrees")
plt.ylabel("Count")
plt.show()
plt.close()
```

### Average clustering, transitivity, average shortest path length

```python
print(f"*** Scale-free network ***")
print(f"Average clustering: {nx.average_clustering(ba_network):.3f}")
print(f"Transitivity {nx.transitivity(ba_network):.3f}")
print(f"Average shortest path length: {nx.average_shortest_path_length(ba_network):.3f}")
```

```python
print(f"*** Random network ***")
print(f"Average clustering: {nx.average_clustering(random_network):.3f}")
print(f"Transitivity {nx.transitivity(random_network):.3f}")
print(f"Average shortest path length: {nx.average_shortest_path_length(random_network):.3f}")
```
